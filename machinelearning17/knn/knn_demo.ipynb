```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_blobs  # make_blobs is used for making clustering data
from sklearn.cluster import KMeans
```

```python

centroids = [(-5,-5,6),(5,5,-3),(3.5,2,4),(2.5,-3.4,-4)]
cluster_std = [1,1,1,1]
X,y=make_blobs(n_samples=200,n_features=3,random_state=12,cluster_std=cluster_std,centers=centroids)
```

```python
import plotly.express as px
fig = px.scatter_3d(x=X[:,0], y=X[:,1],z=X[:,2])
fig.show()
```

```python
wcss=[]
for i in range(1,11):
    km=KMeans(n_clusters=i)
    km.fit_predict(X)
    wcss.append(km.inertia_)
```

```python
import matplotlib.pyplot as plt
plt.plot(range(1,11),wcss)
```
