### chat word treatment

```python
# Dictionary of common chat slangs/abbreviations
chat_words = {
    "AFAIK": "As Far As I Know",
    "AFK": "Away From Keyboard",
    "ASAP": "As Soon As Possible",
    "ATK": "At The Keyboard",
    "ATM": "At The Moment",
    "A3": "Anytime, Anywhere, Anyplace",
    "BAK": "Back At Keyboard",
    "BBL": "Be Back Later",
    "BBS": "Be Back Soon",
    "BFN": "Bye For Now",
    "B4N": "Bye For Now",
    "BRB": "Be Right Back",
    "BRT": "Be Right There",
    "BTW": "By The Way",
    "B4": "Before",
    "CU": "See You",
    "CUL8R": "See You Later",
    "CYA": "See You",
    "FAQ": "Frequently Asked Questions",
    "FC": "Fingers Crossed",
    "FWIW": "For What It's Worth",
    "FYI": "For Your Information",
    "G2G": "Got To Go",
    "GN": "Good Night",
    "GR8": "Great",
    "HAND": "Have A Nice Day",
    "IC": "I See",
    "IDK": "I Don't Know",
    "IMHO": "In My Humble Opinion",
    "IMO": "In My Opinion",
    "IMO": "In My Opinion",
    "LOL": "Laughing Out Loud",
    "NP": "No Problem",
    "OMG": "Oh My God",
    "ROFL": "Rolling On The Floor Laughing",
    "TBH": "To Be Honest",
    "THX": "Thanks",
    "TY": "Thank You",
}

# Function to expand chat words
def chat_conversion(text):
    new_text = []
    for word in text.split():
        # Check if uppercase word is in dictionary
        if word.upper() in chat_words:
            new_text.append(chat_words[word.upper()])
        else:
            new_text.append(word)
    return " ".join(new_text)

# Apply to overview or text columns
text_cols = df.select_dtypes(include=['object', 'string']).columns
for col in text_cols:
    df[col] = df[col].astype(str).apply(chat_conversion)
```

```python
import nltk
from nltk.corpus import stopwords

# Download stop words list
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

# Function to remove stop words
def remove_stopwords(text):
    return " ".join([word for word in text.split() if word not in stop_words])

# Apply to overview or all text columns
text_cols = df.select_dtypes(include=['object', 'string']).columns
for col in text_cols:
    df[col] = df[col].astype(str).apply(remove_stopwords)
```

```python
import emoji

# Function to demojize text
def convert_emojis(text):
    return emoji.demojize(text)

# Apply to text columns
text_cols = df.select_dtypes(include=['object', 'string']).columns
for col in text_cols:
    df[col] = df[col].astype(str).apply(convert_emojis)
```

```python
import spacy

nlp = spacy.load('en_core_web_sm')

# Function to tokenize via spaCy
def spacy_tokenize(text):
    doc = nlp(text)
    return [token.text for token in doc]

df['overview_tokens'] = df['overview'].astype(str).apply(spacy_tokenize)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()

def stem_text(text):
    return " ".join([ps.stem(word) for word in text.split()])

# Apply to overview column
df['overview_stemmed'] = df['overview'].astype(str).apply(stem_text)
```

```python
import spacy

nlp = spacy.load('en_core_web_sm')

def spacy_lemmatize(text):
    doc = nlp(text)
    return " ".join([token.lemma_ for token in doc])

df['overview_lemmatized'] = df['overview'].astype(str).apply(spacy_lemmatize)
```
