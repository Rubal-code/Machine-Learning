```python
import graphviz.backend as be
```

```python
from sklearn.datasets import *
from dtreeviz.trees import *
from IPython.display import Image, display_svg, SVG
```

```python

clas = tree.DecisionTreeClassifier(max_depth=2)  
iris = load_iris()

X_train = iris.data
y_train = iris.target
clas.fit(X_train, y_train)
```

```python
from sklearn.tree import plot_tree
plot_tree(clas)
```

# 1) classification

```python
viz = model(clas, 
               X_train,
               y_train,
               feature_names=iris.feature_names, 
               class_names=["setosa", "versicolor", "virginica"])
display_svg(viz.view())
```

# 2) regrssion

```python
regr = tree.DecisionTreeRegressor(max_depth=1)
califorinia = fetch_california_housing()

X_train = califorinia.data
y_train = califorinia.target
regr.fit(X_train, y_train)

viz = model(regr,
               X_train,
               y_train,
               target_name='price',
               feature_names=califorinia.feature_names,
               
              )
display_svg(viz.view())
```
