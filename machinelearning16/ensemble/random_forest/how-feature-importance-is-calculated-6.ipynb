```python
from sklearn.datasets import make_classification
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
X,y = make_classification(n_samples=100, n_classes=2,
                               n_features=2, n_informative=2, n_redundant=0,
                               random_state=0)
```

```python
X,y
```

```python
clf = DecisionTreeClassifier()
```

```python
clf.fit(X,y)
```

```python
from sklearn.tree import plot_tree
plot_tree(clf)
```

```python
clf.feature_importances_   # the sum of the feature importances is 1
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf=RandomForestClassifier(n_estimators=2)
rf.fit(X,y)
```

```python
rf.feature_importances_
```

```python
print(rf.estimators_[0].feature_importances_)
print(rf.estimators_[1].feature_importances_)
print((rf.estimators_[0].feature_importances_[0] + rf.estimators_[1].feature_importances_[0])/2)
```
